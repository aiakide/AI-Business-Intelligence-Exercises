## Concept at a Glance — Die Speicherkisten-Analogie

Stell Dir eine **Variable** wie eine **beschriftete Aufbewahrungskiste** vor. Auf die Kiste
schreibst Du einen Namen (z.B. `schadenshoehen`), und hinein legst Du einen Wert. Später kannst
Du die Kiste am Namen wiedererkennen und ihren Inhalt abrufen — ohne dass Du Dir den Wert selbst
merken musst.

In Python schreibst Du eine solche Zuweisung mit dem Gleichheitszeichen `=`:

```python
schaden = 800
```

Das bedeutet: "Erzeuge eine Kiste namens `schaden` und lege den Wert `800` hinein."

> 💡 **Good to know:**
> Das `=` in Python ist **keine mathematische Gleichung**! Es bedeutet nicht "ist gleich", sondern
> "weise diesem Namen den Wert rechts zu". `schaden = 800` liest sich also als "schaden **wird**
> 800", nicht als "schaden ist gleich 800".

Manchmal brauchst Du nicht nur eine Kiste, sondern **mehrere zusammengehörige Werte** — z.B. alle
sechs gemeldeten Schadenshöhen eines Monats. Dafür gibt es die **Liste**: eine Kiste, die selbst
wieder mehrere Werte in einer festen Reihenfolge enthält, geschrieben mit eckigen Klammern `[]`.

````python
# I DO: So legst Du eine Liste mit unseren 6 Kfz-Schadenshöhen (in EUR) an
# und gibst sie mit print() aus.

schadenshoehen = [800, 1100, 1100, 1400, 2200, 38000]

print(schadenshoehen)
````

## Step-by-Step Breakdown

1. `schadenshoehen = [800, 1100, 1100, 1400, 2200, 38000]`
   Wir erzeugen eine Liste mit sechs Zahlen und speichern sie unter dem Namen `schadenshoehen`.
   Die Reihenfolge bleibt erhalten — die erste Zahl (`800`) ist der erste gemeldete Schaden.
2. `print(schadenshoehen)`
   Die **Funktion** `print()` gibt den Inhalt der Klammer auf dem Bildschirm aus. Eine Funktion ist
   wie ein kleines Küchenrezept: Du gibst ihr eine Zutat (hier: die Liste), sie liefert ein
   Ergebnis (hier: die Textausgabe).

Rohe Ausgaben wie `[800, 1100, ...]` sind aber wenig lesbar für andere Menschen. Deshalb nutzen wir
**f-Strings** — Textbausteine, in die wir Variablen direkt einbetten können. Ein f-String beginnt
mit einem `f` direkt vor den Anführungszeichen; Variablen kommen in geschweifte Klammern `{...}`.

```python
print(f"Unsere Schadenshöhen: {schadenshoehen}")
```

Zusätzlich gibt es zwei nützliche eingebaute Funktionen für Listen:
- `len(liste)` — zählt, wie viele Elemente in der Liste stecken.
- `sum(liste)` — addiert alle Zahlen der Liste auf.

> ⚠️ **Common Pitfall:**
> Vergiss nicht das `f` vor dem Anführungszeichen! `print("Anzahl: {len(schadenshoehen)}")` (ohne
> `f`) gibt den Text `{len(schadenshoehen)}` wortwörtlich aus, statt ihn zu berechnen.

````python
# I DO: f-Strings kombinieren Text und Variablenwerte in einer Zeile

print(f"Unsere Schadenshöhen: {schadenshoehen}")
print(f"Anzahl der Schäden: {len(schadenshoehen)}")
print(f"Gesamtsumme: {sum(schadenshoehen)} EUR")
````

## Module & `import` — fremden Code nutzen

Niemand schreibt jede Funktion selbst neu. **Module** sind fertige Sammlungen von Code, die andere
Menschen bereits geschrieben haben — vergleichbar mit einem Werkzeugkasten, den Du Dir ausleihst,
statt jedes Werkzeug selbst zu schmieden. Mit dem Schlüsselwort `import` machst Du ein Modul in
Deinem Notebook verfügbar.

Es gibt zwei Arten von Modulen:
- **Standard-Library**: wird automatisch mit Python mitgeliefert, z.B. `statistics` (Statistik-Funktionen).
- **Third-Party-Pakete**: müssen separat installiert werden, z.B. `numpy` (numerisches Rechnen).

```python
import statistics
import numpy as np
```

Das `as np` am Ende ist eine **Alias-Konvention**: Du gibst dem Modul `numpy` den Kurznamen `np`,
damit Du ihn später nicht jedes Mal ausschreiben musst. Das ist reine Bequemlichkeit — `np` und
`numpy` meinen exakt dasselbe Modul.

> 💡 **Good to know:**
> `import numpy as np` ist so verbreitet, dass fast jeder Python-Code zur Datenanalyse diesen Alias
> benutzt. Wenn Du fremden Code liest und `np.irgendwas(...)` siehst, weißt Du jetzt: das ist NumPy.

````python
# I DO: Zwei Module importieren und dieselbe Berechnung mit beiden durchführen

import statistics
import numpy as np

schadenshoehen = [800, 1100, 1100, 1400, 2200, 38000]

print(statistics.mean(schadenshoehen))  # Standard-Library
print(np.mean(schadenshoehen))          # Third-Party-Paket
````

## Punkt-Notation lesen: `modul.funktion(argument)`

Fällt Dir das wiederkehrende Muster in `statistics.mean(schadenshoehen)` und
`np.mean(schadenshoehen)` auf? Beide Zeilen folgen exakt demselben Bauplan:

```
modul . funktion ( argument )
```

Das bedeutet: "Rufe die Funktion `funktion` **aus** dem Modul `modul` auf, und übergib ihr
`argument` als Eingabe." Der Punkt `.` ist dabei kein Satzzeichen, sondern **Teil der Syntax** — er
sagt Python: "gehe in dieses Modul hinein und hole dort die Funktion."

> 💡 **Good to know:**
> Diese Punkt-Notation begegnet Dir ab jetzt ständig: `statistics.mean(...)`, `np.array(...)`,
> `df.corr()`, `schaden.hoehe`. Sobald Du das Muster **Objekt-oder-Modul . Eigenschaft-oder-Funktion**
> erkennst, kannst Du fast jede Zeile Python-Code grob entschlüsseln — auch wenn Du die konkrete
> Funktion noch nicht kennst.

Schauen wir uns das Muster noch einmal mit allen drei Lagemaßen an, die Du im nächsten Notebook
(`02_statistik_in_python.ipynb`) im Detail kennenlernst:

```python
print(f"Mittelwert: {statistics.mean(schadenshoehen)}")
print(f"Median: {statistics.median(schadenshoehen)}")
print(f"Modus: {statistics.mode(schadenshoehen)}")
```

**Selbstcheck:** Bevor Du die nächste Zelle ausführst — kannst Du vorhersagen, welche drei Zahlen
ausgegeben werden? (Tipp: Mittelwert ≈ 7433.33, Median = 1250.0, Modus = 1100)

````python
# WE DO: Führe die Zeile aus und vergleiche mit Deiner Vorhersage von oben

print(f"Mittelwert: {statistics.mean(schadenshoehen)}")
print(f"Median: {statistics.median(schadenshoehen)}")
print(f"Modus: {statistics.mode(schadenshoehen)}")
````
