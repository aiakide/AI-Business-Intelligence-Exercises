## Concept at a Glance — Bauplan und ausgefülltes Formular

Eine **Klasse** ist wie eine **leere Versicherungsformular-Vorlage** — sie legt fest, welche
Felder existieren, aber enthält selbst keine konkreten Werte:

- Schadenshöhe: ❓
- Fahreralter: ❓
- Geschlecht: ❓

Ein **Objekt** ist ein **konkret ausgefüllter Vertrag** — eine konkrete Realisierung der Klasse
mit echten Werten:

- Schadenshöhe: **3.500 €**
- Fahreralter: **45 Jahre**
- Geschlecht: **m**

Jeder Schaden in Deinem Portfolio wird ein neues Objekt — alle folgen demselben Bauplan (derselben
Klasse), aber jedes hat seine eigenen konkreten Werte.

> 💡 **Good to know:**
> Eine Klasse existiert immer nur **abstrakt** — sie ist die Vorlage, nicht der Inhalt. Erst wenn
> Du ein Objekt aus ihr erzeugst ("instanziierst"), bekommt sie echte Daten.

## Klassen erstellen — `class` und `__init__`

Eine Klasse wird mit dem Schlüsselwort `class` definiert. Ihre `__init__`-Methode (sprich:
"Init", von "initialisieren") ist der **Konstruktor** — eine Anleitung, wie beim Erzeugen eines
neuen Objekts die Werte in die richtigen Felder eingetragen werden.

```python
class Schaden:
    def __init__(self, hoehe, alter_fahrer, geschlecht):
        self.hoehe = hoehe
        self.alter_fahrer = alter_fahrer
        self.geschlecht = geschlecht
```

- `def __init__(self, ...)` — die Konstruktor-Methode; sie läuft automatisch, sobald Du ein neues
  Objekt erzeugst.
- `self` bezeichnet **das Objekt selbst**, das gerade erzeugt wird. Python übergibt `self`
  automatisch — Du musst es beim Aufruf nie selbst angeben, nur bei der Definition der Methode.
- `self.hoehe = hoehe` speichert den übergebenen Wert `hoehe` dauerhaft **am Objekt** — als
  sogenanntes **Attribut**.

> 💡 **Good to know:**
> `self` ist wie das Wort "mein" auf dem ausgefüllten Formular: `self.hoehe` bedeutet "**meine**
> Schadenshöhe" — bezogen auf genau das Objekt, das gerade angesprochen wird.

````python
# I DO: Klasse definieren und ein Objekt (eine Instanz) erzeugen

class Schaden:
    def __init__(self, hoehe, alter_fahrer, geschlecht):
        self.hoehe = hoehe
        self.alter_fahrer = alter_fahrer
        self.geschlecht = geschlecht

schaden_1 = Schaden(3500, 45, 'm')
print(schaden_1.hoehe)          # 3500
print(schaden_1.alter_fahrer)   # 45
````

`schaden_1 = Schaden(3500, 45, 'm')` erzeugt eine neue Instanz der Klasse `Schaden` und übergibt
die drei Werte an `__init__`. Python setzt `self` dabei automatisch auf `schaden_1` — Du gibst
`self` beim Aufruf **nie** selbst mit an.

## Methoden — der `ist_ausreisser()`-Umbau

So hast Du es in Notebook 4 gemacht — mit einer eigenständigen Funktion, der Du die Schadenshöhe
jedes Mal explizit übergeben musst:

```python
def ist_ausreisser(hoehe, schwelle=10000):
    return hoehe > schwelle

result = ist_ausreisser(schaden_1.hoehe, schwelle=10000)
```

````python
# I DO: "Vorher" - die klassische Funktion, Schadenshöhe muss durchgereicht werden

def ist_ausreisser(hoehe, schwelle=10000):
    return hoehe > schwelle

result = ist_ausreisser(schaden_1.hoehe, schwelle=10000)
print(result)  # False
````

Der Nachteil: Du musst `schaden_1.hoehe` bei jedem Aufruf explizit durchreichen. Jetzt bauen wir
dieselbe Logik zu einer **Methode** der Klasse `Schaden` um — sie gehört dann zum Objekt und kennt
`self.hoehe` bereits von selbst.

```python
class Schaden:
    def __init__(self, hoehe, alter_fahrer, geschlecht):
        self.hoehe = hoehe
        self.alter_fahrer = alter_fahrer
        self.geschlecht = geschlecht

    def ist_ausreisser(self, schwelle=10000):
        return self.hoehe > schwelle
```

````python
# I DO: "Nachher" - dieselbe Logik als Methode der Klasse

class Schaden:
    def __init__(self, hoehe, alter_fahrer, geschlecht):
        self.hoehe = hoehe
        self.alter_fahrer = alter_fahrer
        self.geschlecht = geschlecht

    def ist_ausreisser(self, schwelle=10000):
        return self.hoehe > schwelle

schaden_1 = Schaden(3500, 45, 'm')
print(schaden_1.ist_ausreisser())  # False - self.hoehe wird automatisch verwendet
````

## Step-by-Step Breakdown

Der entscheidende Unterschied: `schaden_1.ist_ausreisser()` musst Du **keine** Schadenshöhe mehr
übergeben — die Methode kennt `self.hoehe` bereits, weil sie **zum Objekt gehört**. Wenn Du
`schaden_1.ist_ausreisser()` aufrufst, übergibt Python automatisch `schaden_1` als `self` — genau
wie beim Konstruktor.

Das ist der Kern von OOP: **Daten und Verhalten zusammenbündeln**. Statt Werte überall
durchzuschleifen (wie bei reinen Funktionen), fragst Du das Objekt direkt: "Bist du ein
Ausreißer?"

> ⚠️ **Common Pitfall:**
> Vergisst Du `self` als ersten Parameter einer Methode (`def ist_ausreisser(schwelle=10000):` statt
> `def ist_ausreisser(self, schwelle=10000):`), meldet Python einen Fehler — Python übergibt das
> Objekt automatisch als ersten Parameter, und die Methode muss diesen Platz reservieren.

## Schäden als Objekte — eine Liste von Instanzen

Jetzt bauen wir mehrere Schadenshöhen aus unserem bekannten 6er-Portfolio als Liste von Objekten:

````python
# I DO: Liste von Schaden-Objekten erzeugen und filtern

# Schadenshöhen: 800, 1.100, 1.100, 1.400, 2.200, 38.000 EUR
schaeden = [
    Schaden(800, 35, 'w'), Schaden(1100, 50, 'm'),
    Schaden(1100, 42, 'w'), Schaden(1400, 55, 'm'),
    Schaden(2200, 45, 'w'), Schaden(38000, 28, 'm')
]

ausreisser = [s for s in schaeden if s.ist_ausreisser()]
print(len(ausreisser))          # 1 - genau der 38.000-EUR-Schaden
print(ausreisser[0].hoehe)      # 38000
````

Objekte bündeln Daten und Verhalten zusammen. Statt die Schadenshöhe überall durchzureichen,
fragst Du jetzt einfach `schaden.ist_ausreisser()` — lesbar und wartbar, auch bei 400.000
Verträgen.

## Summary & Key Takeaways

- Eine **Klasse** ist der Bauplan (`class Schaden: ...`); ein **Objekt** ist eine konkrete Instanz
  mit echten Werten (`Schaden(3500, 45, 'm')`).
- `__init__(self, ...)` ist der Konstruktor: er setzt die Attribute des neuen Objekts.
- `self` bezeichnet das Objekt selbst; Python übergibt es beim Methodenaufruf automatisch.
- Eine **Methode** (`def methodenname(self, ...): ...`) ist wie eine Funktion, kennt aber bereits
  die Daten des Objekts über `self` — kein manuelles Durchreichen mehr nötig.
- Objekte bündeln Daten (Attribute) und Verhalten (Methoden) an einem Ort — das macht Code bei
  großen Datenmengen lesbarer und wartbarer.
