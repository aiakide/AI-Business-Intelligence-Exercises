## Concept at a Glance — Die Weggabelung

Stell Dir `if`/`elif`/`else` wie eine **Weggabelung auf einer Wanderung** vor: An jeder Kreuzung
prüfst Du eine Bedingung ("Ist der Weg links steiler als 20 %?") und entscheidest je nach Antwort,
welchen Pfad Du nimmst. Python prüft Bedingungen in genau dieser Reihenfolge — von oben nach unten
— und folgt dem ersten Pfad, dessen Bedingung `True` (wahr) ist.

```python
if bedingung_1:
    # nur ausgeführt, wenn bedingung_1 wahr ist
elif bedingung_2:
    # nur ausgeführt, wenn bedingung_1 falsch war UND bedingung_2 wahr ist
else:
    # nur ausgeführt, wenn keine der obigen Bedingungen wahr war
```

> ⚠️ **Common Pitfall:**
> Vergiss nicht den Doppelpunkt `:` am Ende jeder `if`/`elif`/`else`-Zeile! Ohne ihn weiß Python
> nicht, wo die Bedingung endet und der auszuführende Code beginnt. Ebenso wichtig: die
> **Einrückung** (meist 4 Leerzeichen) — sie zeigt Python, welcher Code zu welcher Bedingung gehört.

````python
# I DO: Ist der 45.000-EUR-Schaden ein Ausreißer?
# Schwelle = Mittelwert + 2,5 * Standardabweichung = 7400 + 2.5 * 19100 = 55150 EUR

claim = 45000
mean_claim = 7400
stdev_claim = 19100
threshold = mean_claim + 2.5 * stdev_claim

if claim > threshold:
    print(f"{claim} EUR — Ausreißer erkannt! (Schwelle: {threshold:.0f} EUR)")
elif claim > 0:
    print(f"{claim} EUR — Normaler Schaden (unter der Schwelle von {threshold:.0f} EUR)")
else:
    print("Ungültiger Schadenswert")
````

## Step-by-Step Breakdown

1. `threshold = mean_claim + 2.5 * stdev_claim` berechnet die obere Grenze: 55.150 EUR.
2. `if claim > threshold:` prüft zuerst die erste Bedingung. 45.000 EUR liegt **unter** 55.150 EUR
   — die Bedingung ist `False`, also überspringt Python diesen Block.
3. `elif claim > 0:` wird als nächstes geprüft. 45.000 EUR ist größer als 0 — diese Bedingung ist
   `True`, also wird genau dieser Block ausgeführt.
4. Der `else`-Block wird **gar nicht mehr geprüft**, da schon eine frühere Bedingung zutraf.

**Ergebnis:** 45.000 EUR liegt unter der Schwelle (55.150 EUR) → kein Ausreißer. Das ist ein
undramatisches, aber korrektes Ergebnis — nicht jede Regel muss "Alarm" auslösen, um richtig zu
sein.

## `for`-Schleife — über die bekannten Schadenshöhen

Statt einen einzelnen Schaden zu prüfen, wenden wir die Regel auf **alle** 6 Werte aus unserem
Portfolio an. Die `for`-Schleife läuft eine Liste Element für Element durch — wie eine To-do-Liste,
die Du Punkt für Punkt abarbeitest.

```python
for claim in claims:
    # dieser Code wird für JEDES Element in claims einmal ausgeführt
```

````python
# I DO: for-Schleife über alle 6 Schadenshöhen

claims = [800, 1100, 1100, 1400, 2200, 38000]
threshold = 10000

for claim in claims:
    if claim > threshold:
        print(f"{claim} EUR — Ausreißer!")
    else:
        print(f"{claim} EUR — Normal")
````

Die Schleife führt denselben `if`/`else`-Code 6 Mal aus — einmal für jeden Wert in `claims`, von
vorne nach hinten. Nur der letzte Wert (38.000 EUR) überschreitet die Schwelle von 10.000 EUR.

> 💡 **Good to know:**
> **for vs. while:** Die `for`-Schleife kennt die Anzahl der Durchläufe im Voraus — hier genau 6,
> weil die Liste 6 Elemente hat. Sie eignet sich für "gehe jedes Element einer bekannten Liste
> durch". Die `while`-Schleife (gleich) eignet sich dagegen für "wiederhole, bis eine Bedingung
> nicht mehr erfüllt ist" — die Anzahl der Durchläufe ist vorher nicht bekannt.

## `while`-Schleife — bis ein Ziel erreicht ist

Neues Szenario: Der Versicherer hat eine **Prämienreserve** von 10.000 EUR für neue
Schadenzahlungen. Die Schäden werden nacheinander bearbeitet, **bis die Reserve aufgebraucht ist**
— wir wissen vorher nicht, nach wie vielen Schäden das passiert.

```python
while index < len(claims):
    # wird wiederholt, solange die Bedingung wahr ist
```

Mit `break` können wir eine Schleife jederzeit sicher **abbrechen**, sobald ein Grenzfall erreicht
ist — hier: sobald die Reserve für den nächsten Schaden nicht mehr ausreicht.

````python
# I DO: while-Schleife mit Budget-Grenze und break

claims = [800, 1100, 1100, 1400, 2200, 38000]
budget = 10000
total = 0
index = 0

while index < len(claims):
    if total + claims[index] <= budget:
        total += claims[index]
        print(f"Zahle {claims[index]} EUR (Restbudget danach: {budget - total} EUR)")
        index += 1
    else:
        print(f"Stopp: {claims[index]} EUR würde das Budget sprengen — Reserve reicht nicht mehr.")
        break
````

## Step-by-Step Breakdown — while & break

1. `while index < len(claims):` — solange noch nicht alle 6 Schäden geprüft wurden, wiederhole.
2. `if total + claims[index] <= budget:` — passt der nächste Schaden noch ins Restbudget?
3. Falls ja: zahlen, `total` erhöhen, `index` um 1 weiterschalten (`index += 1`) — sonst würden wir
   ewig den gleichen Schaden prüfen!
4. Falls nein: eine Stopp-Meldung ausgeben und mit `break` die Schleife sofort **verlassen** — auch
   wenn `index < len(claims)` noch zutrifft.

> ⚠️ **Common Pitfall:**
> Vergisst Du `index += 1` in einer `while`-Schleife, bleibt `index` für immer gleich — die
> Bedingung wird nie `False`, und Du hast eine **Endlosschleife** gebaut. Das `break` schützt hier
> zusätzlich davor, dass wir versehentlich mit `claims[index]` außerhalb der Liste zugreifen
> (ein `IndexError`).

## Die Kopier-Falle

Du hast gerade dieselbe Ausreißer-Prüfung `if claim > threshold: ...` in **zwei verschiedenen
Zellen** oben geschrieben. Stell Dir vor, Du brauchst diese Logik noch öfter — beim nächsten
Portfolio, bei der Betrugserkennung, bei Naturschadenanalysen. Was machst Du dann? Vermutlich denselben
Code **kopieren und einfügen**, überall wo Du ihn brauchst.

Das führt zu zwei Problemen:
- **Fehleranfälligkeit:** Änderst Du die Schwelle von 10.000 auf 12.000 EUR, musst Du *jede einzelne
  Kopie* anpassen. Vergisst Du eine Stelle, arbeiten Teile Deines Programms mit veralteten Regeln.
- **Wartungsalptraum:** Denselben Code an 10 Stellen zu pflegen, ist auf Dauer unmöglich.

**Lösung:** Die Logik einmal definieren, überall wiederverwenden. Das ist das Konzept der
**Funktion** — Thema des nächsten Notebooks.

## Summary & Key Takeaways

- `if`/`elif`/`else` prüft Bedingungen der Reihe nach von oben nach unten und führt nur den ersten
  zutreffenden Block aus. Doppelpunkt `:` und Einrückung sind Pflicht.
- `for element in liste:` durchläuft jedes Element einer bekannten Liste genau einmal — die Anzahl
  der Durchläufe steht vorher fest.
- `while bedingung:` wiederholt, solange die Bedingung wahr ist — die Anzahl der Durchläufe ist
  vorher unbekannt. `break` bricht die Schleife sicher ab.
- Wiederholt kopierter `if`/`elif`-Code ("die Kopier-Falle") führt zu Fehleranfälligkeit und
  Wartungsalpträumen.
