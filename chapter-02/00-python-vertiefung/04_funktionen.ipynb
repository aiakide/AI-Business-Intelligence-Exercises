# 4. Funktionen — Logik einmal definieren, überall nutzen

hast dieselbe Ausreißer-Prüfung `if claim > threshold: ...` mehrfach von Hand geschrieben. Jetzt
lernst Du das Werkzeug, das dieses Problem löst: die **Funktion**.

## Concept at a Glance — Das Rezept

Eine **Funktion** ist wie ein **Küchenrezept**: Du definierst es einmal ("Nimm Mehl, Eier, Milch,
rühre um, backe 20 Minuten"), und danach kannst Du es beliebig oft **aufrufen** — jedes Mal mit
anderen Zutaten (**Parametern**), aber immer demselben Ablauf. Das Ergebnis (der fertige Kuchen)
entspricht dem `return`-Wert einer Funktion.

```python
def ist_ausreisser(schaden, schwelle=10000):
    return schaden > schwelle
```

- `def ist_ausreisser(...)` — das Schlüsselwort `def` ("define") plus der Funktionsname.
- `schaden` — ein **erforderlicher Parameter**: ohne ihn kannst Du die Funktion nicht aufrufen.
- `schwelle=10000` — ein **optionaler Parameter mit Default-Wert**: rufst Du die Funktion ohne
  eigene Schwelle auf, wird automatisch 10.000 verwendet.
- `return schaden > schwelle` — liefert das Ergebnis (`True` oder `False`) an den Aufrufer zurück.

> 💡 **Good to know:**
> `return` beendet die Funktion sofort und gibt den Wert dahinter zurück. Ohne `return` würde die
> Funktion zwar ausgeführt, aber kein nutzbares Ergebnis liefern (`None`).

````python
# I DO: Die Ausreißer-Prüfung aus Notebook 3 als Funktion definieren

def ist_ausreisser(schaden, schwelle=10000):
    return schaden > schwelle

print(ist_ausreisser(38000))                    # nutzt den Default: schwelle=10000 -> True
print(ist_ausreisser(38000, schwelle=50000))     # eigene Schwelle übergeben -> False
````

## Step-by-Step Breakdown

1. `ist_ausreisser(38000)` — wir übergeben nur den erforderlichen Parameter `schaden`. Python
   nutzt automatisch den Default-Wert `schwelle=10000`. Da 38.000 > 10.000, lautet das Ergebnis
   `True`.
2. `ist_ausreisser(38000, schwelle=50000)` — wir überschreiben den Default und übergeben explizit
   `schwelle=50000`. Da 38.000 nicht größer als 50.000 ist, lautet das Ergebnis `False`.

**Der entscheidende Vorteil gegenüber der Kopier-Falle:** Ändert sich morgen die Schwelle im
gesamten Unternehmen, passt Du **eine einzige Stelle** an — die Funktionsdefinition — statt jede
Kopie im Code einzeln zu suchen.

> ⚠️ **Common Pitfall:**
> Parameter mit Default-Wert (`schwelle=10000`) müssen in der Parameterliste **nach** den
> erforderlichen Parametern stehen. `def f(schwelle=10000, schaden): ...` würde einen Fehler
> auslösen.

## Funktion + Schleife kombinieren

Jetzt wenden wir `ist_ausreisser()` auf **alle 6 Schadenshöhen** unseres Portfolios an. Es gibt
zwei gleichwertige Wege:

**Weg 1 — die klassische `for`-Schleife** (ausführlich, aber immer verständlich):
```python
ausreisser = []
for s in schadenshoehen:
    if ist_ausreisser(s):
        ausreisser.append(s)
```

**Weg 2 — die List Comprehension** (eine elegante Kurzform derselben Logik):
```python
ausreisser = [s for s in schadenshoehen if ist_ausreisser(s)]
```

Beide Wege liefern exakt dasselbe Ergebnis. Die List Comprehension ist beliebt bei
erfahrenen Python-Entwicklern, weil sie kompakter ist — aber sie ist nur eine **Kurzschreibweise**
für die klassische Schleife, kein neues Konzept. Solange Du unsicher bist, greif ruhig zur
`for`-Schleife — sie ist genauso richtig und oft leichter zu lesen.

````python
# I DO: Weg 1 - klassische for-Schleife

schadenshoehen = [800, 1100, 1100, 1400, 2200, 38000]

ausreisser = []
for s in schadenshoehen:
    if ist_ausreisser(s):
        ausreisser.append(s)

print(ausreisser)  # [38000]
````

````python
# I DO: Weg 2 - dieselbe Logik als List Comprehension (Kurzform)

ausreisser_kurz = [s for s in schadenshoehen if ist_ausreisser(s)]

print(ausreisser_kurz)  # [38000] -- identisch zu Weg 1
````

Das Ergebnis ist in beiden Fällen `[38000]` — die Funktion `ist_ausreisser()` wird jetzt überall
wiederverwendet, statt an mehreren Stellen kopiert zu werden. Ändert sich die Schwelle, passt Du
sie an **einer Stelle** an: in der Funktionsdefinition.

## Guided Practice — Andere Schwelle über die Funktion

## Summary & Key Takeaways

- Eine **Funktion** (`def name(parameter): ... return ...`) bündelt Logik an einer Stelle, statt
  sie im Code zu kopieren — das löst die Kopier-Falle.
- Parameter **ohne** Default-Wert sind erforderlich; Parameter **mit** Default-Wert (z.B.
  `schwelle=10000`) sind optional und können beim Aufruf überschrieben werden.
- `return` liefert das Ergebnis der Funktion an den Aufrufer zurück.
- Funktion + Schleife lassen sich kombinieren, entweder als klassische `for`-Schleife oder als
  kompaktere List Comprehension — beide sind gleichwertig, die `for`-Schleife ist der verständlichere
  Standardweg.
