## Concept at a Glance — Der Werkzeug-Stapel

Stell Dir die drei Bibliotheken als aufeinander aufbauende Stockwerke vor:

- **NumPy** ist das Fundament: extrem schnelle Berechnungen auf ganzen Zahlenreihen (Arrays).
- **Pandas** baut auf NumPy auf und macht die Daten **benutzerfreundlicher** — mit beschrifteten
  Tabellen (DataFrames) statt roher Zahlenreihen.
- **Matplotlib** macht die Ergebnisse **sichtbar** — als Diagramme statt nackter Zahlen.

Du kennst bereits die `for`-Schleife aus Notebook 3: Sie läuft Wert für Wert durch eine Liste und
rechnet nacheinander. NumPy-Arrays rechnen dagegen **alle Werte auf einmal**, in optimiertem
Hintergrund-Code statt einer Python-Schleife. Das nennt man **Vektorisierung** — das Ergebnis ist
identisch, nur (bei großen Datenmengen) 10-100× schneller.

````python
# I DO: NumPy-Array statt Liste - dieselbe Berechnung, jetzt vektorisiert

import numpy as np

schaeden_array = np.array([800, 1100, 1100, 1400, 2200, 38000])
print(schaeden_array.mean())  # 7433.33 - ein NumPy-Array kennt Operationen wie .mean() direkt
````

Statt `for schaden in liste: summe += schaden` zu schreiben, sagst Du dem Array einfach "berechne
Deinen Durchschnitt selbst" — mit `.mean()`. NumPy liefert die Geschwindigkeit; Pandas (gleich)
baut genau darauf auf und macht sie zusätzlich benutzerfreundlich.

## Pandas Series — der Payoff-Moment

Das ist der Payoff-Moment des gesamten Python-Blocks: Dieselben 6 Schadenshöhen wie in Notebook 2,
aber jetzt mit **Pandas**. Vergleiche die beiden Versionen direkt:

```python
# Alt (Notebook 2): drei einzelne Funktionsaufrufe aus dem statistics-Modul
from statistics import mean, median, stdev
print(mean(data), median(data), stdev(data))

# Neu (Pandas): eine Datenstruktur, die ihre Statistik-Funktionen direkt mitbringt
s = pd.Series(data)
print(s.mean(), s.median(), s.std())
```

Eine `pd.Series` ist wie eine NumPy-Array-Version einer Liste, aber mit eingebauten
Statistik-Methoden — Du "fragst die Daten direkt", statt eine externe Funktion aufzurufen.

````python
# I DO: Alte Version (statistics) vs. neue Version (pandas) direkt vergleichen

from statistics import mean, median, stdev
import pandas as pd

data = [800, 1100, 1100, 1400, 2200, 38000]

# Alt: drei Funktionsaufrufe, Modul 11a-Stil
print("statistics:", mean(data), median(data), stdev(data))

# Neu: eine Zeile, Daten "wissen" ihre eigene Statistik
s = pd.Series(data)
print("pandas:    ", s.mean(), s.median(), s.std())
````

Derselbe Schaden-Datensatz, dieselbe Statistik — identisch zu Deiner Handrechnung aus Notebook 2
(7433.33, 1250.0, 15000.0). Der Punkt hier ist nicht Geschwindigkeit; es ist die **Klarheit** und
**Wiederverwendbarkeit** einer Datenstruktur, die ihre Operationen direkt mitbringt.

## Pandas DataFrame — `.corr()`

Ein `pd.DataFrame` ist wie eine Excel-Tabelle innerhalb von Python — mehrere beschriftete Spalten,
die zusammengehören. Wir bauen den Fahrzeugalter/Reparaturkosten-Datensatz aus Notebook 2 als
DataFrame nach:

```python
df = pd.DataFrame({
    'Fahrzeugalter': [1, 2, 3, 4],
    'Reparaturkosten': [200, 300, 500, 600]
})
df.corr()
```

`.corr()` berechnet die Korrelation **zwischen allen Spalten des DataFrames** auf einmal — kein
manuelles `np.corrcoef(x, y)` mehr nötig.

````python
# I DO: DataFrame anlegen und Korrelation mit .corr() berechnen

df = pd.DataFrame({
    'Fahrzeugalter': [1, 2, 3, 4],          # Jahre
    'Reparaturkosten': [200, 300, 500, 600]  # EUR
})

print(df.corr())
````

r ≈ 0,99 — exakt der Wert aus Deinen Handrechnungen und aus Notebook 2, jetzt in einer Zeile
bestätigt statt mühsam mit `np.corrcoef()` berechnet.

## Matplotlib — Korrelation sichtbar machen

Zahlentabellen sind korrekt, aber ein Bild ist oft schneller zu erfassen. `plt.scatter(x, y)`
zeichnet für jedes Wertepaar einen Punkt in ein Koordinatensystem.

```python
import matplotlib.pyplot as plt
plt.scatter(df['Fahrzeugalter'], df['Reparaturkosten'])
plt.xlabel('Fahrzeugalter (Jahre)')
plt.ylabel('Reparaturkosten (EUR)')
plt.show()
```

````python
# I DO: Streudiagramm zeichnen

import matplotlib.pyplot as plt

plt.figure(figsize=(5, 4))
plt.scatter(df['Fahrzeugalter'], df['Reparaturkosten'])
plt.xlabel('Fahrzeugalter (Jahre)')
plt.ylabel('Reparaturkosten (EUR)')
plt.title('Fahrzeugalter vs. Reparaturkosten')
plt.show()
````

Die vier Punkte liegen fast auf einer geraden Linie — r ≈ 0,99 wird sofort visuell klar, ganz ohne
Zahlen lesen zu müssen.
